```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from sklearn.preprocessing import MinMaxScaler, LabelEncoder
import seaborn as sns

def data_read(file_path):
    data = pd.read_csv(file_path)
    display (data.head())
    display(data.info())
    display(data.shape)
    return data

def cleansing(data):
    kolom_teks = data.select_dtypes(include='object').columns
    for c in kolom_teks:
        data[c] = data[c].astype("string").str.strip().str.lower()

    kolom_number = data.select_dtypes(include="number").columns
    kolom_number = list(kolom_number)
    kolom_number.remove('symboling')
    kolom_number.remove('diesel')
    kolom_number.remove('gas')

    for c in kolom_number:
        data[c] = data[c].fillna(data[c].median())

    for c in kolom_teks:
        data[c] = data[c].fillna(data[c].mode()[0])

    data['transaction_date'] = pd.to_datetime(data['transaction_date'], dayfirst=True, format='mixed')
    data = data.drop_duplicates()
        
    return data

scaler = MinMaxScaler()
label_column = ["aspiration", "num-of-doors","engine-location","horsepower-binned",'num-of-cylinders']
le = LabelEncoder()
one_hot = ["body-style","drive-wheels"]
freq_encoding = ["make","engine-type","fuel-system"]

def transformation(data):
    kolom_number = data.select_dtypes(include="number").columns
    kolom_number = list(kolom_number)
    kolom_number.remove('symboling')
    kolom_number.remove('diesel')
    kolom_number.remove('gas')
    data[kolom_number] = scaler.fit_transform(data[kolom_number])
    
    for c in label_column:
        data[c] = le.fit_transform(data[c])

    data = pd.get_dummies(data, columns=one_hot)

    for col in freq_encoding:
        freq = data[col].value_counts(normalize=True)
        data[col] = data[col].map(freq)

    return data

def save_data(data_raw):
  data_raw.to_csv(r"../data/processed/preprocessed_data.csv",index=False)

def main(filename):
    data = data_read(filename)
    data = cleansing(data)
    data = transformation(data)
    save_data(data)
```

```python
main("../data/raw/automobileEDA_dirty_training.csv")
```
